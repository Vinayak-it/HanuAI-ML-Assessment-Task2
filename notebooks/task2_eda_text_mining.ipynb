```python

# Step 4.2: Define keywords for vehicle issue extraction

issue_keywords = {
    "Radio": [
        "radio", "radio module", "radio screen"
    ],
    "Display/Screen": [
        "screen", "display", "black screen", "blank screen"
    ],
    "OnStar/Communication": [
        "onstar", "communication", "connectivity"
    ],
    "Software/Firmware": [
        "software", "firmware", "programming", "update"
    ],
    "Electrical/Wiring": [
        "wiring", "ground", "voltage", "electrical", "connector"
    ],
    "Audio/Speaker": [
        "speaker", "audio", "sound", "volume", "no sound"
    ],
    "Navigation": [
        "navigation", "gps", "map"
    ],
    "Module Failure": [
        "module internal failure", "internal failure",
        "module failed", "module failure"
    ]
}

print("Number of issue categories:", len(issue_keywords))
```

```python

# Step 4.3: Extract issue tags from customer, cause and repair text

import re

def extract_issue_tags(row):
    # Combine the three narrative fields into one searchable string
    text = " ".join(
        str(row[column])
        for column in [
            "CUSTOMER_VERBATIM",
            "CAUSAL_VERBATIM",
            "CORRECTION_VERBATIM"
        ]
        if pd.notna(row[column])
    ).lower()

    tags = []

    # Assign every category whose keyword appears in the text
    for category, keywords in issue_keywords.items():
        if any(
            re.search(r"\b" + re.escape(keyword) + r"\b", text)
            for keyword in keywords
        ):
            tags.append(category)

    return tags


# Keep the original columns unchanged
df["Extracted Issue Tags"] = df.apply(extract_issue_tags, axis=1)

# Display sample results
display(
    df[
        [
            "CUSTOMER_VERBATIM",
            "CAUSAL_VERBATIM",
            "CORRECTION_VERBATIM",
            "Extracted Issue Tags"
        ]
    ].head(10)
)
```

```python

# Step 5.1: Prepare text for clustering

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans

text_columns = [
    "CUSTOMER_VERBATIM",
    "CAUSAL_VERBATIM",
    "CORRECTION_VERBATIM"
]

# Combine complaint, cause and correction descriptions
df["Combined Text"] = (
    df[text_columns]
    .fillna("")
    .astype(str)
    .agg(" ".join, axis=1)
    .str.lower()
    .str.strip()
)

# Identify records with usable text
usable_text = df["Combined Text"].str.len() > 0

print("Total records:", len(df))
print("Records with usable text:", usable_text.sum())
print("Records without usable text:", (~usable_text).sum())

# Keep the original dataset intact
cluster_text = df.loc[usable_text, "Combined Text"]
```

```python

# Step 5.2: Convert text into numerical TF-IDF features

vectorizer = TfidfVectorizer(
    stop_words="english",
    max_features=3000,
    ngram_range=(1, 2),
    min_df=2,
    max_df=0.95
)

X = vectorizer.fit_transform(cluster_text)

print("TF-IDF matrix shape:", X.shape)
print("Number of text features:", len(vectorizer.get_feature_names_out()))
```

```python

# Step 5.3: Group similar vehicle issue descriptions

kmeans = KMeans(
    n_clusters=5,
    random_state=42,
    n_init=10
)

cluster_labels = kmeans.fit_predict(X)

# Assign cluster labels to the matching original records
df["Cluster"] = pd.Series(
    pd.NA,
    index=df.index,
    dtype="Int64"
)

df.loc[cluster_text.index, "Cluster"] = cluster_labels

print("Records assigned to clusters:")
display(df["Cluster"].value_counts().sort_index())
```

```python

# Step 5.4: Display the most important terms in each cluster

terms = vectorizer.get_feature_names_out()
order = kmeans.cluster_centers_.argsort(axis=1)[:, ::-1]

for cluster_id in range(kmeans.n_clusters):
    top_terms = [terms[i] for i in order[cluster_id, :10]]

    print(f"\nCluster {cluster_id}")
    print("Important terms:", ", ".join(top_terms))

    # Show up to three actual complaint descriptions
    examples = df.loc[
        df["Cluster"] == cluster_id,
        "CUSTOMER_VERBATIM"
    ].head(3)

    for example in examples:
        print("-", example)
```

```python

# Step 5.5: Compare clusters with the original complaint categories

cluster_comparison = pd.crosstab(
    df["Cluster"],
    df["COMPLAINT_CD_DESC"]
)

display(cluster_comparison)

# Show the three most common complaint categories in each cluster
for cluster_id in sorted(df["Cluster"].dropna().unique()):
    print(f"\nCluster {cluster_id}")

    top_categories = (
        df.loc[df["Cluster"] == cluster_id, "COMPLAINT_CD_DESC"]
        .value_counts()
        .head(3)
    )

    display(top_categories)
```

### Interpretation of Text Clustering Results

TF-IDF vectorization and KMeans clustering were used to group similar vehicle complaint, cause, and repair narratives into five exploratory clusters. The clusters help identify recurring themes involving radio and audio performance, display or screen issues, navigation, connectivity, camera/video functionality, and software or module-related repairs.

These groups can help stakeholders identify recurring patterns and prioritize further investigation. However, clusters represent textual similarities rather than confirmed technical root causes. Further validation using repair records and engineering investigation is required to establish the underlying causes.

```python

# Step 7.2: Separate symptoms, diagnosed causes and repair actions

import re

def find_categories(text, keyword_rules):
    if pd.isna(text):
        return []

    text = str(text).lower()
    matched = []

    for category, keywords in keyword_rules.items():
        if any(
            re.search(r"\b" + re.escape(keyword) + r"\b", text)
            for keyword in keywords
        ):
            matched.append(category)

    return matched


# Customer-reported symptoms
symptom_rules = {
    "Blank/Black Screen": ["blank screen", "black screen", "screen just black"],
    "No Audio": ["no sound", "no audio", "no sound from radio"],
    "Inoperative Radio": ["radio is inoperable", "radio does not turn on", "radio will not come on"],
    "Navigation Problem": ["navigation", "navi", "maps"],
    "Camera Problem": ["backup camera", "rear camera", "camera inop"],
    "Connectivity Problem": ["onstar", "bluetooth", "wifi", "wi-fi"]
}

# Technician-described causes
cause_rules = {
    "Internal Module Failure": ["internal failure", "module internal failure", "internal fault"],
    "Logic Lock": ["logic lock"],
    "Software/Programming Issue": ["software failure", "software issue", "programming issue"],
    "Electrical/Wiring Issue": ["wiring fault", "poor ground", "voltage issue"],
    "Cause Not Confirmed": ["unable to duplicate", "could not duplicate", "could not verify"]
}

# Repair actions
repair_rules = {
    "Radio Replacement": ["replaced the radio", "installed a new radio", "replace radio"],
    "Module Replacement": ["replaced the radio module", "replace radio module"],
    "Software Update": ["updated radio software", "software update"],
    "Programming": ["programmed radio", "programmed the radio"],
    "Reset/Restart": ["logic reset", "reset the radio"]
}

df["Symptom Tags"] = df["CUSTOMER_VERBATIM"].apply(
    lambda text: find_categories(text, symptom_rules)
)

df["Cause Tags"] = df["CAUSAL_VERBATIM"].apply(
    lambda text: find_categories(text, cause_rules)
)

df["Repair Action Tags"] = df["CORRECTION_VERBATIM"].apply(
    lambda text: find_categories(text, repair_rules)
)

display(
    df[
        [
            "Event id",
            "CUSTOMER_VERBATIM",
            "CAUSAL_VERBATIM",
            "CORRECTION_VERBATIM",
            "Symptom Tags",
            "Cause Tags",
            "Repair Action Tags"
        ]
    ].sample(10, random_state=42)
)
```

```python
# Step 7.5: Improve symptom, cause, and repair extraction

import re
import pandas as pd

# Match multiple common expressions in the service narratives
symptom_rules = {
    "Blank/Black Screen": [
        r"\bblank screen\b",
        r"\bblack screen\b",
        r"\bscreen.{0,30}\bblack\b",
        r"\bscreen.{0,30}\bblank\b"
    ],
    "No Audio": [
        r"\bno sound\b",
        r"\bno audio\b",
        r"\baudio.{0,20}\binop(?:erative)?\b",
        r"\bsound.{0,20}\binop(?:erative)?\b"
    ],
    "Inoperative Radio": [
        r"\bradio.{0,30}\binop(?:erative)?\b",
        r"\bradio.{0,30}\bnot working\b",
        r"\bradio.{0,30}\bwill not turn on\b",
        r"\bradio.{0,30}\bdoes not turn on\b",
        r"\bradio.{0,30}\bintermittently inop\b"
    ],
    "Navigation Problem": [
        r"\bnavigation\b",
        r"\bnavi\b",
        r"\bmaps?\b",
        r"\bgps\b"
    ],
    "Camera Problem": [
        r"\bbackup camera\b",
        r"\brear camera\b",
        r"\bcamera.{0,20}\binop(?:erative)?\b"
    ],
    "Connectivity Problem": [
        r"\bonstar\b",
        r"\bbluetooth\b",
        r"\bwi[\s-]?fi\b",
        r"\bconnectivity\b",
        r"\bdisconnect(?:s|ed|ing)?\b"
    ]
}

cause_rules = {
    "Internal Module Failure": [
        r"\binternal.{0,25}\bfailure\b",
        r"\bmodule.{0,25}\binternal failure\b",
        r"\binternal.{0,25}\bmalfunction\b",
        r"\bmodule.{0,25}\bmalfunction(?:ing)?\b",
        r"\bmalfunctioning radio\b",
        r"\bfault internal to the radio\b"
    ],
    "Logic Lock": [
        r"\blogic lock\b"
    ],
    "Software/Programming Issue": [
        r"\bsoftware failure\b",
        r"\bsoftware issue\b",
        r"\bprogramming issue\b",
        r"\bprogramming error\b",
        r"\bsoftware error\b"
    ],
    "Electrical/Wiring Issue": [
        r"\bwiring fault\b",
        r"\bpoor ground\b",
        r"\bvoltage issue\b",
        r"\belectrical fault\b",
        r"\bopen circuit\b",
        r"\bshort circuit\b"
    ],
    "Cause Not Confirmed": [
        r"\bunable to duplicate\b",
        r"\bcould not duplicate\b",
        r"\bcould not verify\b",
        r"\bno fault found\b",
        r"\bunable to confirm\b"
    ]
}

repair_rules = {
    "Radio Replacement": [
        r"\breplac(?:e|ed|ing)\b.{0,50}\bradio\b",
        r"\bnew radio\b",
        r"\bradio replacement\b"
    ],
    "Module Replacement": [
        r"\breplac(?:e|ed|ing)\b.{0,50}\bradio module\b",
        r"\breplac(?:e|ed|ing)\b.{0,50}\bmodule\b"
    ],
    "Software Update": [
        r"\bupdated radio software\b",
        r"\bsoftware update\b",
        r"\bsoftware was updated\b",
        r"\bperformed update\b"
    ],
    "Programming": [
        r"\bprogrammed\b.{0,40}\bradio\b",
        r"\bprogramming\b.{0,40}\bradio\b",
        r"\bradio\b.{0,40}\bprogrammed\b"
    ],
    "Reset/Restart": [
        r"\blogic reset\b",
        r"\breset the radio\b",
        r"\breset power to\b.{0,30}\bradio\b",
        r"\bglobal systems reset\b"
    ]
}


def extract_pattern_tags(text, rules):
    if pd.isna(text):
        return []

    text = str(text).lower()
    tags = []

    for category, patterns in rules.items():
        if any(re.search(pattern, text) for pattern in patterns):
            tags.append(category)

    return tags


# Recalculate tags using the updated rules
df["Symptom Tags"] = df["CUSTOMER_VERBATIM"].apply(
    lambda text: extract_pattern_tags(text, symptom_rules)
)

df["Cause Tags"] = df["CAUSAL_VERBATIM"].apply(
    lambda text: extract_pattern_tags(text, cause_rules)
)

df["Repair Action Tags"] = df["CORRECTION_VERBATIM"].apply(
    lambda text: extract_pattern_tags(text, repair_rules)
)

print("Updated tags successfully.")
print("Total records:", len(df))

display(
    df[
        [
            "Event id",
            "Symptom Tags",
            "Cause Tags",
            "Repair Action Tags"
        ]
    ].head(10)
)
```

```python
# Step 7.9: Avoid classifying a radio-module-only replacement
# as a complete radio replacement.

def extract_refined_repair_tags(text):
    if pd.isna(text):
        return []

    text = str(text).lower()
    tags = []

    module_replacement = bool(re.search(
        r"\breplac(?:e|ed|ing)\b.{0,40}\bradio module\b",
        text
    ))

    screen_only = bool(re.search(
        r"\breplac(?:e|ed|ing)\b.{0,30}\b(?:radio )?(?:display )?screen\b",
        text
    ))

    radio_replacement_patterns = [
        r"\breplaced radio\b",
        r"\breplace radio\b",
        r"\breplacing radio\b",
        r"\breplaced the radio\b",
        r"\breplace the radio\b",
        r"\bnew radio\b",
        r"\breplaced radio assembly\b",
        r"\breplaced radio chassis\b"
    ]

    if (
        not module_replacement
        and not screen_only
        and any(re.search(pattern, text)
                for pattern in radio_replacement_patterns)
    ):
        tags.append("Radio Replacement")

    if module_replacement:
        tags.append("Module Replacement")

    if re.search(
        r"\bprogrammed\b.{0,40}\bradio\b|\bradio\b.{0,40}\bprogrammed\b",
        text
    ):
        tags.append("Programming")

    if re.search(
        r"\b(?:updated radio software|software update|performed update)\b",
        text
    ):
        tags.append("Software Update")

    if re.search(
        r"\b(?:logic reset|reset the radio|global systems reset)\b",
        text
    ):
        tags.append("Reset/Restart")

    return tags


df["Repair Action Tags"] = df["CORRECTION_VERBATIM"].apply(
    extract_refined_repair_tags
)

print("Records with repair tags:",
      df["Repair Action Tags"].apply(lambda tags: len(tags) > 0).sum())

display(
    df.loc[
        df["Event id"].isin(["D-15303", "D-14862", "D-15039"]),
        ["Event id", "CORRECTION_VERBATIM", "Repair Action Tags"]
    ]
)
```

## 9.2 Root-Cause Hypotheses and Business Recommendations

### Key Findings

1. **Complaint concentration:** Audio (337 records), Other infotainment issues (263), and Communication (194) account for 794 of the 1,000 records (79.4%). These categories represent the largest areas for further investigation.

2. **Frequently extracted symptoms:** Blank/black screen was detected in 149 records (14.9%), followed by inoperative radio in 86 records (8.6%), connectivity problems in 67 records (6.7%), and no audio in 66 records (6.6%). These are keyword-extracted symptom categories and may overlap.

3. **Recorded cause descriptions:** Internal module failure was detected by the keyword rules in 111 records (11.1%). This is a text-extraction result, not an independently verified failure rate.

4. **Repair-action patterns:** The refined keyword rules detected radio replacement in 594 records (59.4%) and programming in 331 records (33.1%). These categories can overlap. The matches indicate that replacement and programming are prominent in the repair narratives; they do not establish repair necessity or repair success.

5. **Text clustering:** Five exploratory clusters were produced. Cluster sizes were 359, 46, 97, 377, and 121 records. The clusters help organize similar narratives, but their themes should not be treated as confirmed root causes.

### Root-Cause Hypotheses for Further Investigation

- **Radio or display hardware:** The frequency of blank-screen, inoperative-radio, and internal-module-failure language suggests investigating radio/display hardware and related connections.
- **Software or programming:** Programming and software-related language suggests checking software versions, programming history, and whether repeated visits follow updates or reprogramming.
- **Connectivity and communication:** Connectivity complaints suggest investigating communication-module interactions, network connections, and conditions associated with connection failures.
- **Intermittent faults:** Narratives describing symptoms that cannot be duplicated during inspection suggest that diagnostic procedures may need to capture intermittent conditions and relevant diagnostic-code history.

These are hypotheses for investigation, not proven causes. Confirmation requires reviewing individual service records and technical evidence.

### Business Recommendations

1. **Prioritize high-volume complaint categories.** Begin with Audio, Other infotainment issues, and Communication because together they represent 79.4% of the dataset.
2. **Review blank-screen cases.** Examine the 149 records tagged with blank/black-screen symptoms and compare their diagnosed causes, repair actions, vehicle models, and dates.
3. **Audit repeated replacement and programming cases.** Check whether vehicles return with similar symptoms after radio replacement or programming, using event and repair history where available.
4. **Improve diagnostic documentation.** Encourage consistent recording of confirmed causes, diagnostic codes, tests performed, parts replaced, and whether the complaint was resolved.
5. **Validate automated text tags.** Manually review a sample of tagged and untagged records and measure extraction precision and recall before using the tags for operational decisions.
6. **Investigate temporal patterns.** Compare complaint volumes over time while accounting for partial months and any available vehicle-population or sales-volume data.

### Limitations

- The analysis covers 1,000 supplied records and may not represent the wider vehicle population.
- Keyword-based tagging can miss alternative wording and may classify a mention as an action or cause even when the narrative describes an attempt, recommendation, or historical event.
- Multiple tags can apply to one record; therefore, category counts and percentages are not mutually exclusive.
- Clustering is exploratory and does not establish causal relationships.
- The analysis does not independently verify repair success, recurrence, or the effectiveness of recommended interventions.

### Conclusion

The supplied records are concentrated in infotainment-related complaint categories, particularly Audio, Other infotainment issues, and Communication. Blank-screen symptoms and radio replacement are prominent in the extracted narratives. These patterns provide a practical basis for targeted record review, better diagnostic documentation, and further investigation of hardware, software, and connectivity issues. All root-cause conclusions should be validated against the underlying technical evidence before operational decisions are made.

```python

def normalize_text(text):
    if pd.isna(text):
        return ""

    text = str(text).replace("\xa0", " ").replace("\u200b", "")
    text = re.sub(r"\s+", " ", text)

    return text.strip().lower()


def find_categories(text, keyword_rules):
    text = normalize_text(text)

    if not text:
        return []

    matched = []

    for category, keywords in keyword_rules.items():
        for keyword in keywords:
            normalized_keyword = normalize_text(keyword)

            if re.search(
                r"\b" + re.escape(normalized_keyword) + r"\b",
                text
            ):
                matched.append(category)
                break

    return matched
```

```python

def extract_final_repair_tags(text):
    text = normalize_text(text)

    if not text:
        return []

    tags = []

    # Detect replacement of the radio module.
    module_replacement = bool(re.search(
        r"\b(?:replac(?:e|ed|ing)|removed and replaced)\b"
        r".{0,50}\bradio module\b",
        text
    ))

    # Detect replacement of the display screen only.
    screen_only = bool(re.search(
        r"\b(?:replac(?:e|ed|ing))\b"
        r".{0,40}\b(?:radio )?(?:display )?screen\b",
        text
    ))

    # Detect replacement of the radio itself.
    radio_replacement_patterns = [
        r"\breplac(?:e|ed|ing)\s+(?:the\s+)?radio\b",
        r"\bnew radio\b",
        r"\breplac(?:e|ed|ing)\s+radio assembly\b",
        r"\breplac(?:e|ed|ing)\s+radio chassis\b",
        r"\bordered radio and installed\b",
    ]

    if (
        not module_replacement
        and not screen_only
        and any(re.search(p, text) for p in radio_replacement_patterns)
    ):
        tags.append("Radio Replacement")

    if module_replacement:
        tags.append("Module Replacement")

    # Programming and common spelling variations.
    programming_patterns = [
        r"\bprogram(?:med|ming)?\b.{0,60}\bradio\b",
        r"\bradio\b.{0,60}\bprogram(?:med|ming)?\b",
        r"\bprogram(?:med|ming)?\s+via\s+(?:sps|usb)\b",
        r"\bprogr?amming\b",
        r"\bprogramming for updated software\b",
        r"\bradiomodule reprogramming\b",
    ]

    if any(re.search(p, text) for p in programming_patterns):
        tags.append("Programming")

    # Software updates.
    if re.search(
        r"\b(?:updated radio software|software update|"
        r"performed update|radio update|update via usb)\b",
        text
    ):
        tags.append("Software Update")

    # Reset/restart actions.
    if re.search(
        r"\b(?:logic reset|reset the radio|reset radio|"
        r"global systems reset)\b",
        text
    ):
        tags.append("Reset/Restart")

    return tags
```

```python

def extract_final_repair_tags(text):
    text = normalize_text(text)

    if not text:
        return []

    tags = []

    # Identify module replacement.
    module_replacement = bool(re.search(
        r"\b(?:replac(?:e|ed|ing)|removed and replaced)\b"
        r".{0,50}\bradio module\b",
        text
    ))

    # Identify display/screen-only replacement.
    screen_only = bool(re.search(
        r"\b(?:replac(?:e|ed|ing))\b"
        r".{0,40}\b(?:radio )?(?:display )?screen\b",
        text
    ))

    # Identify full radio replacement.
    radio_replacement_patterns = [
        r"\breplac(?:e|ed|ing)\s+(?:the\s+)?radio\b",
        r"\bnew radio\b",
        r"\breplac(?:e|ed|ing)\s+radio assembly\b",
        r"\breplac(?:e|ed|ing)\s+radio chassis\b",
        r"\bordered radio and installed\b",
        r"\binstalled new radio\b",
    ]

    if (
        not module_replacement
        and not screen_only
        and any(re.search(p, text) for p in radio_replacement_patterns)
    ):
        tags.append("Radio Replacement")

    if module_replacement:
        tags.append("Module Replacement")

    # Require a clear programming-action phrase.
    programming_patterns = [
        r"\bprogrammed\s+(?:the\s+)?radio\b",
        r"\bprogrammed\s+radio\s+(?:using|via|with)\b",
        r"\bprogram(?:med|ed)\s+via\s+(?:sps|usb)\b",
        r"\bprogramming\s+(?:the\s+)?radio\b",
        r"\bprogramming\s+for\s+(?:the\s+)?radio\b",
        r"\bperformed programming\b",
        r"\bprogram and set up radio\b",
        r"\breprogrammed\s+(?:the\s+)?radio\b",
        r"\bradio module reprogramming\b",
        r"\bprogrammed radio receiver\b",
        r"\bprogramming radio using\b",
        r"\bprogramming radio via\b",
        r"\bprogamming\b.{0,40}\b(?:radio|software|usb)\b",
        r"\b(?:replaced|installed)\b.{0,60}\bprogrammed\b",
    ]

    if any(re.search(p, text) for p in programming_patterns):
        tags.append("Programming")

    # Software updates: keep this separate from programming.
    software_update_patterns = [
        r"\bupdated radio software\b",
        r"\bsoftware update\b",
        r"\bperformed update\b",
        r"\bradio update via\b",
        r"\bupdate via usb\b",
    ]

    if any(re.search(p, text) for p in software_update_patterns):
        tags.append("Software Update")

    # Reset/restart.
    if re.search(
        r"\b(?:logic reset|reset the radio|reset radio|"
        r"global systems reset)\b",
        text
    ):
        tags.append("Reset/Restart")

    return tags
```
